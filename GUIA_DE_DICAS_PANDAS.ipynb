```python
dropna()          # Remover valores nulos
fillna()          # Preencher valores nulos
replace()         # Substituir valores específicos
astype()          # Alterar tipo de dados
str.strip()       # Remover espaços de strings
str.lower() / str.upper()   # Padronizar texto
duplicated() / drop_duplicates()  # Remover duplicadas
isna() / notna()  # Verificar valores nulos
rename()          # Renomear colunas
```

```python
apply()           # Aplicar funções personalizadas
map()             # Transformação simples de colunas
lambda            # Operações rápidas (junto com apply/map)
groupby()         # Agrupar dados
pivot_table()     # Criar tabelas dinâmicas
melt()            # “Desempilhar” (transformar colunas em linhas)
agg()             # Agregar dados com múltiplas funções
sort_values()     # Ordenar dados
reset_index()     # Resetar índices
set_index()       # Definir coluna como índice
```
